## 🇹🇿 Choosing a Tanzanian problem

Pick something you find genuinely interesting — you will work on it for the whole session. It must relate to a **Tanzanian use case** in **any sector**. Some starting ideas:

**🌾 Agriculture**
- Predict **maize/rice yield** from rainfall, soil, and inputs (regression).
- Classify **crop disease** from tabular sensor or survey data (classification).
- Segment **smallholder farmers** by practices and outputs (clustering).

**🏥 Health**
- Predict **malaria / disease risk** from clinic and environmental data (classification).
- Model **child nutrition / stunting** indicators from DHS-style data (regression/classification).
- Forecast **clinic attendance** or vaccine uptake.

**💰 Finance / Economy**
- **Financial inclusion** — who is likely to be unbanked? (classification — like the worked example).
- **Mobile-money** adoption or transaction patterns.
- **Loan default** or credit risk.

**🚍 Other sectors**
- **Transport**: predict daladala/bus demand or travel time.
- **Education**: predict student performance or dropout risk.
- **Energy**: model household electricity access or consumption.
- **Tourism**: forecast visitor numbers to national parks.

### 📂 Where to find data

- **Tanzania National Bureau of Statistics** (nbs.go.tz).
- **World Bank Open Data**, **WHO**, **FAO**, **DHS Program**.
- **Zindi** (zindi.africa) — African data-science competitions with local datasets.
- **Your own data** — a survey, scraped public data, or records you have access to.

## 🐙 Git & GitHub workflow (required)

Version control is a core professional skill. You must develop your project in a Git repository and push it to GitHub.

### One-time setup
```bash
# configure your identity (once per machine)
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
```

### Starting your project
```bash
# create a folder and initialise git
mkdir tz-capstone && cd tz-capstone
git init

# create the recommended structure (see below), then:
git add .
git commit -m "Initial project structure"

# create an empty repo on github.com, then link and push:
git remote add origin https://github.com/<your-username>/tz-capstone.git
git branch -M main
git push -u origin main
```

### Working day-to-day — commit often, with clear messages
```bash
git add notebooks/analysis.ipynb
git commit -m "Add EDA section with distribution plots"
git push
```

> 💡 **Good commit habits:** commit after each meaningful step (e.g. "clean missing values", "add baseline model", "tune random forest"). Small, descriptive commits tell the story of your work and are part of the assessment.

### A clean `.gitignore`
Create a file named `.gitignore` so you don't commit clutter:
```
__pycache__/
.ipynb_checkpoints/
*.pyc
.DS_Store
venv/
.env
```

## 🗂️ Recommended repository structure

A clean, conventional layout makes your project easy to navigate and shows professionalism:

```
tz-capstone/
├── README.md                 <- project summary + how to run (REQUIRED)
├── requirements.txt          <- libraries needed (pandas, scikit-learn, ...)
├── .gitignore
├── data/
│   ├── raw/                  <- original, untouched dataset
│   └── processed/            <- cleaned data your code produces
├── notebooks/
│   └── analysis.ipynb        <- your main, runnable analysis
├── src/                      <- optional: reusable .py helper scripts
│   └── preprocessing.py
└── reports/
    └── figures/              <- saved plots for the README/report
```

> You don't need every folder — but a `README.md`, a `notebooks/` analysis, and a `data/` (or a link to the data) are the minimum. Two starter files are provided alongside this guide: **`capstone_template.ipynb`** (a skeleton to fill in) and **`README_template.md`**.

## ✅ Assessment rubric

Your project will be assessed on the full process, not just the final accuracy number. A model that scores modestly but is **rigorously and honestly built** beats a high score with leakage or no documentation.

| Area | Weight | What we look for |
|------|:------:|------------------|
| **Problem framing** | 10% | Clear, relevant Tanzanian problem; correct task type (classification/regression/clustering). |
| **Data handling** | 15% | Sensible cleaning; missing values, types, and duplicates handled; no data leakage. |
| **EDA** | 15% | Meaningful summaries and visualisations; insights stated, not just plots dumped. |
| **Feature work** | 10% | Thoughtful feature selection and any engineering, with justification. |
| **Modelling & validation** | 25% | Appropriate model(s); train/test split or CV; tuning; **correct metrics**; overfitting checked. |
| **Results & communication** | 15% | Clear conclusions, honest about limitations, actionable insights. |
| **Repository & reproducibility** | 10% | Runnable top-to-bottom; clean structure; good README; sensible Git commit history. |

### 🏁 Definition of "done"
- [ ] The notebook **runs top-to-bottom without errors** (Kernel → Restart & Run All).
- [ ] EDA, modelling, and **clear conclusions** are all present.
- [ ] Metrics are appropriate (e.g. **F1/recall for imbalanced** classification; **RMSE/R²** for regression).
- [ ] No data leakage (transforms fit inside pipelines / on training data only).
- [ ] A **README** explains the problem, data, approach, key results, and how to run it.
- [ ] The repository is pushed to GitHub with a **sensible commit history**.

## 🗓️ Suggested time plan

Use the session to make strong progress; you may finish documentation afterwards.

1. **Define & find data (15%)** — lock your problem and dataset early; don't over-search.
2. **Clean & EDA (30%)** — understand the data deeply before modelling.
3. **Feature work & modelling (30%)** — baseline first, then improve and tune.
4. **Validation & honest evaluation (10%)** — CV, correct metrics, overfitting/leakage check.
5. **Document & push (15%)** — write the README/report, tidy the notebook, commit and push.

> ⚠️ **Common pitfalls to avoid** (all covered in earlier sessions):
> - **Data leakage** — fit scalers/encoders inside a `Pipeline`, never on the whole dataset before splitting.
> - **Wrong metric** — accuracy lies on imbalanced data; use **F1/recall**. (Recall the FinScope "accuracy trap".)
> - **No baseline** — always start simple, then justify added complexity.
> - **Plot dumping** — every chart should support a stated insight.
> - **Last-minute commit** — commit throughout, not just once at the end.
